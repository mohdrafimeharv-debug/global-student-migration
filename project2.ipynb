```python
d=d.fillna({"placement_country":"unknown","placement_company":"unknow","language_proficiency_test":"unknoen"})
```

```python
d
```

```python
import matplotlib.pyplot as plt
```

```python
#top 10 destination countries
plt.figure(figsize=(10, 5))

d["destination_country"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Destination Countries")
plt.xlabel("Destination country")
plt.ylabel("Number of Students")
plt.legend()
```

```python
#reason for student migration
d["enrollment_reason"].value_counts().plot(kind="bar")
plt.show()
```

```python
#top 10 origin countries
d["origin_country"].value_counts().head(10).plot(kind="bar")
plt.show()
```

```python
#GPA/score of placed student
d[d["placement_status"]=="Placed"]["gpa_or_score"].plot(kind="hist")
plt.show() 
```

```python
#GPA/score of not placed student
d[d["placement_status"]=="Not Placed"]["gpa_or_score"].plot(kind="hist")
plt.show()
```

```python
#average starting salary by placement country-it helps student understand differences in potential earnings between countries
d.groupby("placement_country")["starting_salary_usd"].mean().nlargest(10).plot(kind="bar")
plt.show()
```

```python
#average starting salary by university-it help compare careee outcomes associated with different unversities
d.groupby("university_name")["starting_salary_usd"].mean().nlargest(10).plot(kind="bar")
plt.show()
```

```python
#average starting salary by field of study-it helps students explore which fields offer higher average starting salaries in the data set
d.groupby("field_of_study")["starting_salary_usd"].mean().nlargest(10).plot(kind="bar")
plt.show()
```

```python
#scholarship by orgin country-it helps us understand scholarship distribution among students from different countries
pd.crosstab(d["origin_country"], d["scholarship_received"]).head(10).plot(kind="bar")
plt.show()
```
